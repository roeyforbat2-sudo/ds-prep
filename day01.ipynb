##מסכם

```python
def gc_content(seq, decimals=2):
    gc = seq.count("G") + seq.count("C")
    return round(gc / len(seq) * 100, decimals)

print(gc_content("AAAA"))
```

```python
def reverse_complement(seq):
    """Return the reverse complement of a DNA sequence."""
    complement = {"A": "T", "T": "A", "G": "C", "C": "G"}
    result = ""
    for base in seq:
        result += complement[base]   # בונים את הרצף המשלים
    return result[::-1]              # הופכים אותו ומחזירים

reverse_complement("ATGCGTAC")   # 'GTACGCAT'
```
