```python
seq1 = "ATATAT"
seq2 = "AAAAAAAA"
seq3 = "ACGTACGTACGT"
seq4 = "GATTACAGATTACA"
def unique_kmers(seq, k):
    kmers = set()
    for i in range(len(seq) - k + 1):
        kmer = seq[i:i+k]
        kmers.add(kmer)
    return len(kmers)
# בדיקות: אם הפונקציה נכונה, כל השורות ידפיסו True
print(unique_kmers(seq1, 2) == 2)    # AT, TA
print(unique_kmers(seq2, 3) == 1)    # AAA
print(unique_kmers(seq3, 4) == 4)    # ACGT, CGTA, GTAC, TACG
print(unique_kmers(seq4, 3) == 7)
print(unique_kmers(seq4, 1) == 4)    # A, C, G, T
```
